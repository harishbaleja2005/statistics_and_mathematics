```python
#1.Create a Python list to represent a playlist of 5 song durations in seconds (e.g., [210, 180, 245, 200, 175]).
#  Print the total duration of the playlist.

playlist = [210, 180, 245, 200, 175]

total_duration = sum(playlist)

print("playlist:", playlist)
print("total duration:", total_duration, "seconds")
```

```python
#2.Represent a 3x3 matrix in Python to store the ratings (1-5) given by 3 friends to 3 different food items 
# on Swiggy. Print the matrix and display the rating each friend gave to the second food item.

# 3x3 matrix of ratings
ratings = [
    [5, 4, 3],   # Friend 1
    [4, 5, 2],   # Friend 2
    [3, 4, 5]    # Friend 3
]

# print the complete matrix
print("Ratings Matrix:")
for row in ratings:
    print(row)

# print the rating each friend gave to the second food item
print("\nRatings for the second food item:")

for i in range(3):
    print(f"friend {i+1}: {ratings[i][1]}")
```

```python
#3.Given a matrix representing the number of orders for 4 different dishes across 3 days (rows: days, columns: dishes),
# write code to calculate and print the total orders for each dish.

# Matrix representing orders
# Rows = Days
# Columns = Dishes

orders = [
    [20, 15, 18, 10],   # Day 1
    [25, 12, 20, 14],   # Day 2
    [22, 18, 16, 12]    # Day 3
]

# print the matrix
print("Orders Matrix:")
for row in orders:
    print(row)

# calculate total orders for each dish
print("total orders for each dish:")

for dish in range(4):
    total = 0
    for day in range(3):
        total += orders[day][dish]
    print(f"Dish {dish + 1}: {total}")
```

```python
#4.Imagine you have a table of 5 Flipkart products with columns for price and quantity sold. Represent this 
# data as a matrix in Python, then write a function to calculate the total revenue for each product.<br><br>
# <em><strong>Hint:</strong> Multiply price and quantity for each row, then print the results.</em>

# Matrix representing Flipkart products
# Each row = [Price, Quantity Sold]

products = [
    [500, 20],    # Product 1
    [750, 15],    # Product 2
    [300, 40],    # Product 3
    [1200, 10],   # Product 4
    [900, 25]     # Product 5
]

# Function to calculate total revenue
def calculate_revenue(products):
    print("revenue for each product:")

    for i in range(len(products)):
        price = products[i][0]
        quantity = products[i][1]

        revenue = price * quantity

        print(f"product {i+1}: ₹{revenue}")

# call the function
calculate_revenue(products)
```
