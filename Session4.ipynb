```python
#1.Given the daily order counts for a Swiggy delivery partner over 10 days: [12, 15, 17, 14, 13, 16, 200, 18, 14, 15], 
#plot a simple graph (hand-drawn or using any tool) and describe whether the data is left skewed, right skewed, or symmetrical. Explain your reasoning based on the shape.
import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]
days = range(1, 11)

plt.plot(days, orders, marker='o')
plt.title("Daily Swiggy Orders")
plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.grid(True)

plt.show()

# description:
# most daily order counts are between 12 and 18, but one unusually high value (200)
# makes the distribution right skewed (positively skewed).
```

```python
#2.Take the following list of Instagram influencer follower counts (in thousands): [5, 7, 8, 8, 9, 10, 12, 15, 95]. 
#Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.
# <br><br><em><strong>Hint:</strong> If the mean is pulled away from the median, the data is skewed in that direction.</em>
import numpy as np

followers=[5, 7, 8, 8, 9, 10, 12, 15, 95]

mean=np.mean(followers)
median=np.median(followers)

print("mean is:",mean)
print("median is:",median)

if mean>median:
    print("data is right skewed (positively skewed)")
elif mean<median:
    print("data is left skewed (negatively skewed)")
else:
    print("data is symmetrical")
```
