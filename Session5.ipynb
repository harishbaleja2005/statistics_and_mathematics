```python
#2.Write a Python script that simulates random sampling: given a list of 1000 fake user IDs (e.g., user1 to user1000), 
# select 50 random users to simulate a survey sample. Print the selected user IDs.<br><br><em><strong>Hint:</strong> Use the random.sample() function from Python's random module.</em>
import random

population=[]

# create a population of 1000 fake user IDs
for i in range(1,1001):
    population.append(f"user:{i}")

# randomly select 50 users as the survey sample
sample=random.sample(population,50)

#print the selected users
for i1 in sample:
    print(i1)
```

```python
#3.Imagine you are analyzing food delivery times on Swiggy. Explain in 2-3 sentences why you 
# would use sampling instead of collecting data from every single order ever made.

import random

all_orders = [
    25, 30, 28, 35, 22, 40, 27, 31, 29, 33,
    26, 24, 38, 32, 21, 36, 34, 23, 37, 39
]

sample_orders = random.sample(all_orders, 5)

print("population (all orders):", all_orders)
print("sample (5 random orders):", sample_orders)
print("explanation:")
print("sampling is used because collecting data from every Swiggy order is time-consuming and expensive.")
print("a randomly selected sample gives a good representation of the overall delivery times while saving time and resources.")

# description:
# swiggy has millions of food delivery orders
# collecting data from every single order (the population)
# would be time-consuming and expensive
# instead, a random sample of orders is selected because
# it is faster and less costly to analyze
# if the sample is chosen randomly, it can accurately
# represent the entire population and provide reliable
# insights into food delivery times
```

```python
#4.Given this scenario: A survey about music preferences is only sent to users who recently
# streamed Bollywood songs on Spotify. Identify and explain the type of bias this introduces in the sample.

spotify_users = [
    "Bollywood", "Pop", "Rock", "Hip-Hop", "Classical",
    "Jazz", "EDM", "Bollywood", "Rock", "Pop"
]

biased_sample = [genre for genre in spotify_users if genre == "Bollywood"]

print("population (All Users):", spotify_users)
print("biased Sample:", biased_sample)
print("explantion:")
print("the survey is sent only to users who recently streamed Bollywood songs")
print("this excludes users who prefer other music genres, so the sample does not represent all Spotify users")

# type of bias: selection bias

# description:
# the survey is sent only to users who recently streamed
# bollywood songs on spotify
# users who prefer other music genres such as pop, rock,
# jazz, or classical are not included in the survey
# therefore, the sample does not represent all Spotify users
# this introduces selection bias and the survey results
# may be biased toward Bollywood music preferences
```

```python
#5.Choose any two types of sampling (for example: random sampling, stratified sampling, systematic sampling, cluster sampling).
# For each, describe how you would use it to select a sample of users from BookMyShow to study movie booking habits.

# 1. random sampling

import random

users = [
    "User1", "User2", "User3", "User4", "User5",
    "User6", "User7", "User8", "User9", "User10"
]

sample = random.sample(users, 3)

print("random sampling")
print("population:", users)
print("sample:", sample)

print("explanation:")
print("every user has an equal chance of being selected.")
print("the selected users are used to study movie booking habits.")

# in random sampling, every BookMyShow user has an equal
# chance of being selected.
# for example, if BookMyShow has 1,00,000 users,
# i would randomly select 500 users to study
# their movie booking habits.
# this method reduces bias and gives every user
# an equal opportunity to be included in the sample.


# 2. systematic sampling

users = [
    "User1", "User2", "User3", "User4", "User5",
    "User6", "User7", "User8", "User9", "User10"
]


start = random.randint(0, 1) 
sample = users[start::2]

print("systematic sampling")
print("population:", users)
print("sample:", sample)

print("explanation:")
print("every 2nd user is selected from the list")
print("these users are surveyed to study movie booking habits")

# in systematic sampling, I would arrange all
# bookMyShow users in a list.
# then, i would select every 2nd user
# from the list to form the sample.
# this method is simple, quick, and ensures
# users are selected at regular intervals
# to study movie booking habits.
```
