```python
#1.Write down the null hypothesis (H0) and alternative hypothesis (H1) for the following scenario: You believe that 
# people spend more time on Instagram than on YouTube daily.
from scipy.stats import ttest_rel

instagram = [90, 80, 100, 70, 120, 95, 85, 110, 75, 105]
youtube = [70, 60, 80, 65, 100, 90, 75, 95, 70, 90]

# perform paired t-test
t_stat, p_value = ttest_rel(instagram, youtube)

# convert to one-tailed p-value
p_value = p_value / 2

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("reject H0")
else:
    print("fail to reject H0")
```

```python
#2.Given the data: Group A (users shown a new Zomato home page) had an average order value of ₹350 from 30 
# users, and Group B (old home page) had an average order value of ₹320 from 30 users. Assume both groups 
# have a standard deviation of ₹50. Calculate the p-value using a two-sample t-test (you may use Python, R, 
# or an online calculator), and decide whether to accept or reject the null hypothesis at a 0.05 significance 
# level.<br><br><em><strong>Hint:</strong> Use scipy.stats.ttest_ind in Python or any online t-test calculator if you are not comfortable with code.</em>
from scipy.stats import ttest_ind_from_stats

mean1 = 350
std1 = 50
n1 = 30

mean2 = 320
std2 = 50
n2 = 30

t_stat, p_value = ttest_ind_from_stats(
    mean1, std1, n1,
    mean2, std2, n2,
    equal_var=True
)

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("reject H0")
else:
    print("fail to Reject H0")
```

```python
#3.Pick any feature from your favorite app (like the new dark mode in WhatsApp or a redesigned payment 
# button in Paytm) and describe how you would set up an A/B test for it. Clearly state what your control and 
# variant groups would be, what metric you would measure, and what your null and alternative hypotheses are.
from scipy.stats import ttest_ind

control = [48, 50, 47, 49, 46, 51, 48]
variant = [52, 53, 51, 54, 50, 52, 53]

t_stat, p_value = ttest_ind(variant, control)

print("T-statistic:", t_stat)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("reject the null hypothesis.")
    print("the new dark mode significantly improves engagement.")
else:
    print("fail to reject the null hypothesis.")
    print("no significant improvement found.")
```

```python
#4.Suppose you run an A/B test for a new 'Add to Wishlist' button design on Flipkart. Out of 1000 users, 520 
# clicked the new button versus 480 who clicked the old button. Without doing detailed calculations, explain 
# if this difference is likely to be significant or just due to chance. Briefly explain how the p-value helps you decide.

# ------------------------------------------
# A/B Testing: Flipkart 'Add to Wishlist' Button
# ------------------------------------------

# Scenario:
# Flipkart introduced a new 'Add to Wishlist' button design.
# They want to know whether the new design performs better than the old one.

# Control Group (A):
# Users who see the OLD button.
# Total users = 1000
# Clicks = 480
# Click Rate = 480 / 1000 = 48%

# Variant Group (B):
# Users who see the NEW button.
# Total users = 1000
# Clicks = 520
# Click Rate = 520 / 1000 = 52%

# Difference:
# New button has a 4% higher click rate than the old button.

# ------------------------------------------
# View 1: Business Perspective
# ------------------------------------------

# The company wants more users to add products to their wishlist.
# A higher click rate means users are engaging more with the feature.
# If the increase is real, Flipkart should launch the new design.

# ------------------------------------------
# View 2: Statistical Perspective
# ------------------------------------------

# We cannot say the new design is better just because it has more clicks.
# The difference might simply happen due to random chance.
# Therefore, we perform a statistical hypothesis test.

# Null Hypothesis (H0):
# The new button does NOT improve the click rate.
# Any observed difference is due to random variation.

# Alternative Hypothesis (H1):
# The new button DOES improve the click rate.

# ------------------------------------------
# View 3: What is the p-value?
# ------------------------------------------

# The p-value tells us:
# "If there is actually no difference between the buttons,
# how likely is it that we would observe a difference this large?"

# Small p-value (< 0.05):
# The observed difference is unlikely to be due to chance.
# Reject the null hypothesis.

# Large p-value (>= 0.05):
# The observed difference could easily happen by chance.
# Fail to reject the null hypothesis.

# ------------------------------------------
# View 4: Practical Interpretation
# ------------------------------------------

# Old button click rate = 48%
# New button click rate = 52%

# Since there are 1000 users in each group,
# a 4% increase is noticeable.

# Without calculating the p-value,
# we can only say the difference is LIKELY significant,
# but we cannot be certain.

# ------------------------------------------
# View 5: Final Decision
# ------------------------------------------

# If p-value < 0.05:
#    -> Reject H0
#    -> New button significantly improves clicks.
#    -> Launch the new design.

# If p-value >= 0.05:
#    -> Fail to reject H0
#    -> Difference may simply be due to chance.
#    -> Keep testing or retain the old design.
```
