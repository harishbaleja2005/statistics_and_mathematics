```python
#1.Given an array of daily step counts from your fitness app for one week (e.g., [4500, 7000, 5000, 8000, 4000, 9000, 6000]),
# calculate the variance by hand using the variance formula and show each calculation step.
daily_steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]


mean = sum(daily_steps) / len(daily_steps)

squared_diff = []

for step in daily_steps:
    diff = step - mean
    square = diff ** 2
    squared_diff.append(square)

total = sum(squared_diff)

population_variance = total / len(daily_steps)
sample_variance = total / (len(daily_steps) - 1)
print("mean is:",mean)
print("population variance:", population_variance)
print("sample variance:", sample_variance)
```

```python
#3.Imagine two friends track their daily spending on Zomato over 5 days. Friend A spends [200, 200, 200, 200, 200] 
# and Friend B spends [100, 300, 150, 400, 50]. Calculate the standard deviation for both and explain in 2-3 lines who has more consistent spending and why.
import numpy as np

friend_a=np.array([200, 200, 200, 200, 200])
friend_b=np.array([100, 300, 150, 400, 50])

new_a=[]
new_b=[]
for i in friend_a:
    new_a.append((i-np.mean(friend_a))**2)

for i1 in friend_b:
    new_b.append((i1-np.mean(friend_b))**2)

std_a=sum(new_a)/len(new_a)
std_b=sum(new_b)/len(new_b)

print("friend a standard deviation is:",np.sqrt(std_a))
print("friend b standard deviation is:",np.sqrt(std_b))

print("friend A has more consistent spending because the standard deviation is 0")
print("friend B has a higher standard deviation because the spending amounts vary a lot from day to day")
```

```python
#4.You are given the monthly salary data of 6 employees in a startup: [28000, 29000, 27000, 60000, 26500, 27500]. 
# Calculate the variance and standard deviation, then interpret in 2-3 lines what the high or low spread means for this team.
employees_salary=np.array([28000, 29000, 27000, 60000, 26500, 27500])

new_salary=[]
for i in employees_salary:
    new_salary.append((i-np.mean(employees_salary))**2)

variance=sum(new_salary)/len(new_salary)

print("mean is:",np.mean(employees_salary))
print("variance is:",variance)
print("standard deviation is:",np.sqrt(variance))

print("the standard deviation is high because one employee earns 60000, which is much higher than the other salaries.")
print("this means the salaries are widely spread out, so there is high variation in the team's salaries.")
```
