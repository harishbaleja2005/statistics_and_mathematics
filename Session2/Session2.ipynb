```python
#1.Create a Python script called playlist_stats.py that takes a list of daily Spotify song play counts 
# (e.g., [120, 135, 150, 200, 120, 90, 200]) and calculates the mean (average) number of plays.
import numpy as np

song_play_count=np.array([120, 135, 150, 200, 120, 90, 200])
print("average play is:",np.mean(song_play_count))
```
