```python
#1.Generate 100 random numbers in Python that follow a normal distribution with mean 50 and standard deviation 
# 10 using numpy, then plot the bell curve using matplotlib.

import numpy as np
import matplotlib.pyplot as plt

# generate 100 random numbers with mean = 50 and standard deviation = 10
data = np.random.normal(loc=50, scale=10, size=100)

# plot the bell curve (histogram)
plt.hist(data, bins=10, density=True)

# add title and labels
plt.title("Normal Distribution (Mean=50, SD=10)")
plt.xlabel("Value")
plt.ylabel("Density")

# show the plot
plt.show()

# this program generates 100 random numbers that follow a normal distribution
# with a mean of 50 and a standard deviation of 10 using NumPy.
# it then displays the data as a histogram to visualize the bell-shaped distribution using Matplotlib.
```

```python
#2.Given an array of 20 daily step counts from your fitness tracker app, calculate the mean, median, and mode, 
# and check if they are approximately equal as expected in a normal distribution.<br><br><em><strong>Hint:</strong> Use numpy.mean(), numpy.median(), and scipy.stats.mode().</em>

import numpy as np
from scipy import stats

# array of 20 daily step counts
steps = [
    8000, 8500, 9000, 8700, 9200,
    8800, 9100, 8600, 8900, 9050,
    8950, 8750, 9150, 8850, 9000,
    8900, 9100, 8700, 8800, 9000
]

# calculate mean, median, and mode
mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=True)

print("mean:", mean)
print("median:", median)
print("mode:", mode.mode[0])

# check if they are approximately equal
if abs(mean - median) < 200 and abs(mean - mode.mode[0]) < 200:
    print("The data is approximately normally distributed.")
else:
    print("The data is not approximately normally distributed.")
```

```python
#3.Simulate the scores of 200 players in an online game (e.g., PUBG, Free Fire) using a normal distribution 
# with mean 70 and standard deviation 15, then count how many players scored within one standard deviation (between 55 and 85) 
# of the mean.<br><br><em><strong>Hint:</strong> Use the 68-95-99 rule to check your answer.</em>

import numpy as np

# generate scores of 200 players
scores = np.random.normal(loc=70, scale=15, size=200)

# count players scoring within one standard deviation (55 to 85)
count = np.sum((scores >= 55) & (scores <= 85))

print("total players:", len(scores))
print("players scoring between 55 and 85:", count)
print("percentage:", (count / len(scores)) * 100, "%")
```

```python
#4.Pick any real-world scenario (like delivery times for Zomato orders or playlist song durations on Spotify) 
# and explain in 3-4 lines why the normal distribution might be a good fit for modeling this data.

import numpy as np

# simulate delivery times (in minutes) for 100 Zomato orders
delivery_times = np.random.normal(loc=30, scale=5, size=100)

# calculate average delivery time
mean_time = np.mean(delivery_times)

print("average delivery Time:", round(mean_time, 2), "minutes")
print("sample delivery Times:", delivery_times[:10])

print("explanation:")
print("most delivery times are close to the average of 30 minutes.")
print("only a few orders are much faster or slower.")
print("this creates a bell-shaped pattern, so the normal distribution is a good model.")
```
