```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6, 7]
posts = [1, 2, 2, 3, 4, 5, 6]

plt.scatter(hours, posts)

plt.title("Instagram Hours vs Posts")
plt.xlabel("Hours Spent on Instagram")
plt.ylabel("Number of Posts")

plt.grid(True)

plt.show()

# Observation
print("the scatter plot shows a positive correlation.")
print("as the number of hours spent on Instagram increases,")
print("the number of posts made also generally increases.")
print("this means users who spend more time on Instagram tend to make more posts.")
```

```python
import numpy as np


zomato = [5, 8, 10, 12, 15, 18]
swiggy = [4, 7, 9, 11, 14, 17]

correlation = np.corrcoef(zomato, swiggy)

print("correlation matrix:")
print(correlation)

print("correlation coefficient:", correlation[0,1])

# Observation
print("Observation:")

if correlation[0,1] > 0:
    print("the correlation is positive")
    print("as the number of Zomato orders increases,")
    print("the number of Swiggy orders also generally increases")
    print("this indicates that both variables move in the same direction")
elif correlation[0,1] < 0:
    print("the correlation is negative")
    print("as the number of Zomato orders increases,")
    print("the number of Swiggy orders generally decreases")
    print("this indicates that both variables move in opposite directions")
else:
    print("the correlation is zero.")
    print("there is no clear relationship between the two variables.")
```
