```python
#1.Write a Python function bayes_posterior(prior, likelihood, evidence) that calculates and returns the posterior 
# probability using Bayes theorem formula for given values.

def bayes_posterior(prior, likelihood, evidence):
    posterior = (prior * likelihood) / evidence
    return posterior

# example values
prior = 0.01          # probability of having a disease
likelihood = 0.95     # probability of testing positive if disease is present
evidence = 0.05       # overall probability of testing positive

# calculate posterior probability
result = bayes_posterior(prior, likelihood, evidence)

print("posterior probability:", result)
```

```python
#2.Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants 
# offer a discount, while only 20% of regular restaurants do. If you see a discount, use Bayes theorem in 
# code to find the probability that the restaurant is premium.<br><br><em><strong>Hint:</strong> Define prior, 
# likelihood, and evidence in your code with comments.</em>


# prior: probability that a restaurant is premium
prior = 0.10

# likelihood: probability of discount if restaurant is premium
likelihood = 0.80

# probability that a restaurant is regular
regular = 0.90

# probability of discount if restaurant is regular
discount_given_regular = 0.20

# evidence:
# total probability of seeing a discount
evidence = (likelihood * prior) + (discount_given_regular * regular)

# bayes theorem
posterior = (prior * likelihood) / evidence

print("probability that the restaurant is premium given a discount:", posterior)
print(f"percentage: {posterior * 100:.2f}%")
```

```python
#3.Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the 
# probability that 'free' appears in spam vs non-spam emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.

emails = [
    "Win a FREE iPhone now",
    "Claim your FREE prize",
    "FREE vacation offer",
    "Congratulations! You won",
    "Earn money fast",
    "FREE gift card inside",
    "Limited time FREE deal",
    "Exclusive FREE coupon",
    "Click here for FREE rewards",
    "FREE cashback today",

    "Meeting at 3 PM",
    "Project update",
    "Your electricity bill",
    "Family dinner tonight",
    "Class schedule",
    "Job interview reminder",
    "Bank account statement",
    "Doctor appointment",
    "Happy Birthday!",
    "Team meeting tomorrow"
]

#probability that an email is spam
prior = 0.50      # 10 spam emails out of 20

# likelihood: probability that the word 'free' appears in spam emails
likelihood = 0.80

# probability that 'free' appears in non-spam emails
free_given_nonspam = 0.20

# probability that an email is not spam
nonspam = 1 - prior

# evidence: overall probability of seeing the word 'free'
evidence = (likelihood * prior) + (free_given_nonspam * nonspam)

# bayes theorem
posterior = (prior * likelihood) / evidence

print("probability that an email is spam if it contains 'free':", posterior)
print(f"percentage: {posterior * 100:.2f}%")
```
