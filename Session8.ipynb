```python
#1.Write a Python function calculate_probability(event_count, total_outcomes) that returns the probability 
# of an event, and use it to find the probability of getting 'Heads' when tossing a coin.

def calculate_probability(event_count, total_outcomes):
    probability = event_count / total_outcomes
    return probability

# probability of getting Heads in a coin toss
heads = 1
total_outcomes = 2

result = calculate_probability(heads, total_outcomes)

print("probability of getting Heads:", result)
```

```python
#2.Simulate rolling a six-sided dice 100 times using Python's random module, count how many times a '6' appears, and calculate the experimental 
# probability of getting a '6'.<br><br><em><strong>Hint:</strong> Use random.randint(1,6) in a loop and keep a counter for '6'.</em>
import random

# counter for number of times 6 appears
count_six = 0

# roll the dice 100 times
for i in range(100):
    roll = random.randint(1, 6)  # Random number between 1 and 6

    if roll == 6:
        count_six += 1

# calculate experimental probability
experimental_probability = count_six / 100

print("number of times 6 appeared:", count_six)
print("experimental Probability:", experimental_probability)
```

```python
#3.Given an array of genres from your last 20 Spotify songs (like ['pop', 'rock', 'pop', 'hiphop', ...]), write code to calculate the 
# probability of randomly picking a 'pop' song from your recent history.

# List of genres from the last 20 Spotify songs
genres = [
    "pop", "rock", "pop", "hiphop", "pop",
    "jazz", "rock", "pop", "classical", "pop",
    "hiphop", "rock", "pop", "edm", "pop",
    "country", "pop", "rock", "pop", "jazz"
]

# count the number of pop songs
pop_count = genres.count("pop")

# total number of songs
total_songs = len(genres)

# calculate probability
probability = pop_count / total_songs

# display results
print("number of Pop songs:", pop_count)
print("total songs:", total_songs)
print("probability of picking a Pop song:", probability)
```

```python
#4.Write a function that, given two arrays — one for users who ordered food on Zomato and one for users who 
# ordered dessert — calculates the probability that a randomly selected user ordered both food and dessert (i.e., intersection over total unique users).

def calculate_probability(food_users, dessert_users):
    # users who ordered both food and dessert
    both_users = set(food_users) & set(dessert_users)

    # total unique users
    total_users = set(food_users) | set(dessert_users)

    # probability
    probability = len(both_users) / len(total_users)

    return probability


# example data
food_users = ["Amit", "Riya", "Karan", "Neha", "Rahul"]
dessert_users = ["Riya", "Neha", "Priya", "Rahul", "Ankit"]

# call the function
result = calculate_probability(food_users, dessert_users)

print("Probability of ordering both food and dessert:", result)
```

```python
#5.Explain with a short Python code example how conditional probability could be used to improve recommendations 
# in an app like Flipkart (e.g., probability of buying headphones given user bought a phone).<br><br><em>
# <strong>Hint:</strong> Use counts of users who bought both items vs total who bought a phone.</em>

def conditional_probability(both_bought, phone_buyers):
    return both_bought / phone_buyers

# example data
phone_buyers = 100          # Users who bought a phone
both_bought = 35            # Users who bought both phone and headphones

# calculate probability
probability = conditional_probability(both_bought, phone_buyers)

# display result
print("Probability of buying headphones given a phone was bought:", probability)
```
